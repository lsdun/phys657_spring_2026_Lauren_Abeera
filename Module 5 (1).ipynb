```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, Subset
from torchvision import datasets, transforms
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, accuracy_score
import torch.nn.functional as F

transform = transforms.ToTensor()
mnist_train = datasets.MNIST('./data', train=True,  download=True, transform=transform)
mnist_test  = datasets.MNIST('./data', train=False, download=True, transform=transform)

X_train_np = mnist_train.data.numpy().reshape(-1, 784).astype(np.float32) / 255.0
y_train_np = mnist_train.targets.numpy()
X_test_np  = mnist_test.data.numpy().reshape(-1, 784).astype(np.float32) / 255.0
y_test_np  = mnist_test.targets.numpy()

sub10k_idx   = np.random.choice(len(mnist_train), 10000, replace=False)
sub500_idx   = np.random.choice(len(mnist_train), 500,   replace=False)
train10k     = Subset(mnist_train, sub10k_idx)
train500     = Subset(mnist_train, sub500_idx)

test_loader  = DataLoader(mnist_test,  batch_size=1000, shuffle=False)

# Part (c): Define network 
class MLP(nn.Module):
    def __init__(self, hidden_sizes):
        super().__init__()
        layers = []
        in_size = 784
        for h in hidden_sizes:
            layers.append(nn.Linear(in_size, h))
            layers.append(nn.ReLU())
            in_size = h
        layers.append(nn.Linear(in_size, 10))
        self.net = nn.Sequential(*layers)

    def forward(self, x):
        x = x.view(x.size(0), -1)   # flatten
        return self.net(x)

def train_model(model, train_dataset, n_epochs=20, lr=1e-3, val_dataset=None):
    loader     = DataLoader(train_dataset, batch_size=128, shuffle=True)
    val_loader = DataLoader(val_dataset or mnist_test, batch_size=1000, shuffle=False)
    criterion  = nn.CrossEntropyLoss()
    optimizer  = torch.optim.Adam(model.parameters(), lr=lr)

    train_losses, val_losses = [], []

    for epoch in range(n_epochs):
        model.train()
        epoch_loss = 0
        for X, y in loader:
            optimizer.zero_grad()
            loss = criterion(model(X), y)
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()
        train_losses.append(epoch_loss / len(loader))

        model.eval()
        with torch.no_grad():
            v_loss = sum(criterion(model(X), y).item() for X, y in val_loader)
        val_losses.append(v_loss / len(val_loader))

    return train_losses, val_losses

def get_accuracy(model, loader):
    model.eval()
    correct, total = 0, 0
    with torch.no_grad():
        for X, y in loader:
            preds = model(X).argmax(dim=1)
            correct += (preds == y).sum().item()
            total   += y.size(0)
    return correct / total

#Part (c): Train baseline [128] on full data
model_c = MLP([128])
tr_c, va_c = train_model(model_c, mnist_train, n_epochs=20)
acc_c = get_accuracy(model_c, test_loader)
print(f"Part (c) test accuracy: {acc_c:.4f}")

#Part (d): Architecture comparison (10k subset)
architectures = {
    'Baseline [128]' : [128],
    'Narrow [32]'    : [32],
    'Wide [256]'     : [256],
    'Deep [128,64,32]': [128, 64, 32],
}

results = {}
for name, hidden in architectures.items():
    m = MLP(hidden)
    tr, va = train_model(m, train10k, n_epochs=20)
    acc = get_accuracy(m, test_loader)
    results[name] = dict(train=tr, val=va, acc=acc, model=m)
    print(f"{name}: {acc:.4f}")

# Plot loss curves
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
for name, r in results.items():
    ax1.plot(r['train'], label=name)
    ax2.plot(r['val'],   label=name)
ax1.set_title('Training Loss');  ax1.set_xlabel('Epoch'); ax1.legend()
ax2.set_title('Validation Loss'); ax2.set_xlabel('Epoch'); ax2.legend()
plt.suptitle('Part (d): Architecture Comparison')
plt.tight_layout(); plt.show()

# Accuracy table
print("\nPart (d) Accuracy Table")
print(f"{'Architecture':<20} {'Test Accuracy':>13}")
print("-" * 35)
for name, r in results.items():
    print(f"{name:<20} {r['acc']:>13.4f}")

#Part (e): Overfitting on 500 samples
model_e = MLP([128, 64, 32])
tr_e, va_e = train_model(model_e, train500, n_epochs=50)

plt.figure(figsize=(8, 5))
plt.plot(tr_e, label='Train Loss')
plt.plot(va_e, label='Val Loss')
plt.xlabel('Epoch'); plt.ylabel('Loss')
plt.title('Part (e): Overfitting on 500 samples')
plt.legend(); plt.show()

overfit_epoch = int(np.argmin(va_e))
print(f"Test loss starts increasing around epoch {overfit_epoch}")

#Part (f): Early stopping
def train_early_stopping(model, train_dataset, n_epochs=50, lr=1e-3, patience=5):
    loader     = DataLoader(train_dataset, batch_size=128, shuffle=True)
    val_loader = DataLoader(mnist_test, batch_size=1000, shuffle=False)
    criterion  = nn.CrossEntropyLoss()
    optimizer  = torch.optim.Adam(model.parameters(), lr=lr)

    best_val, best_weights, best_epoch = float('inf'), None, 0
    train_losses, val_losses = [], []
    epochs_no_improve = 0

    for epoch in range(n_epochs):
        model.train()
        epoch_loss = 0
        for X, y in loader:
            optimizer.zero_grad()
            loss = criterion(model(X), y)
            loss.backward()
            optimizer.step()
            epoch_loss += loss.item()
        train_losses.append(epoch_loss / len(loader))

        model.eval()
        with torch.no_grad():
            v_loss = sum(criterion(model(X), y).item() for X, y in val_loader)
        v_loss /= len(val_loader)
        val_losses.append(v_loss)

        if v_loss < best_val:
            best_val     = v_loss
            best_weights = {k: v.clone() for k, v in model.state_dict().items()}
            best_epoch   = epoch
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= patience:
                print(f"Early stopping at epoch {epoch}")
                break

    return train_losses, val_losses, best_weights, best_epoch

model_f_full  = MLP([128, 64, 32])
model_f_early = MLP([128, 64, 32])

train_early_stopping(model_f_full,  train10k, n_epochs=50, patience=999)  # no early stop
tr_f, va_f, best_w, best_ep = train_early_stopping(model_f_early, train10k, n_epochs=50, patience=5)

model_f_early.load_state_dict(best_w)
acc_full  = get_accuracy(model_f_full,  test_loader)
acc_early = get_accuracy(model_f_early, test_loader)
print(f"Full training accuracy:         {acc_full:.4f}")
print(f"Early-stopped accuracy (ep {best_ep}): {acc_early:.4f}")

#Part (g): Visualize learned weights
best_model = results['Baseline [128]']['model']
W = best_model.net[0].weight.data  # shape [128, 784]

fig, axes = plt.subplots(8, 16, figsize=(16, 8))
for i, ax in enumerate(axes.flatten()):
    w = W[i].reshape(28, 28).numpy()
    ax.imshow(w, cmap='RdBu', vmin=-w.abs().max(), vmax=w.abs().max())
    ax.axis('off')
plt.suptitle('Part (g): First-layer weight visualizations')
plt.tight_layout(); plt.show()

#Part (h): Error analysis
model_h = results['Baseline [128]']['model']
model_h.eval()

misclassified = []
with torch.no_grad():
    for X, y in DataLoader(mnist_test, batch_size=1000, shuffle=False):
        logits = model_h(X)
        probs  = F.softmax(logits, dim=1)
        preds  = logits.argmax(dim=1)
        wrong  = (preds != y).nonzero(as_tuple=True)[0]
        for idx in wrong:
            misclassified.append((X[idx], y[idx].item(), preds[idx].item(), probs[idx]))
        if len(misclassified) >= 5:
            break

fig, axes = plt.subplots(5, 2, figsize=(10, 15))
for i, (img, true, pred, prob) in enumerate(misclassified[:5]):
    # Image
    axes[i, 0].imshow(img.squeeze(), cmap='gray')
    axes[i, 0].set_title(f'True: {true}  |  Predicted: {pred}')
    axes[i, 0].axis('off')
    # Probability bar chart
    axes[i, 1].bar(range(10), prob.numpy())
    axes[i, 1].set_xticks(range(10))
    axes[i, 1].set_xlabel('Digit class')
    axes[i, 1].set_ylabel('Probability')
    axes[i, 1].set_title('Class probabilities')

plt.suptitle('Part (h): Misclassified examples', fontsize=14)
plt.tight_layout(); plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

#Setup 
np.random.seed(42)
X = np.linspace(0, 1, 200).reshape(-1, 1)
y = np.sin(2 * np.pi * X).ravel() + 0.2 * np.random.randn(200)

X_tensor = torch.tensor(X, dtype=torch.float32)
y_tensor = torch.tensor(y, dtype=torch.float32)
y_true   = np.sin(2 * np.pi * X.ravel())  # clean sine for plotting

class RegNet(nn.Module):
    def __init__(self, H):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(1, H),
            nn.ReLU(),
            nn.Linear(H, 1)
        )

    def forward(self, x):
        return self.net(x).squeeze(-1)

#Train helper
def train_regnet(H, n_epochs=8000, lr=0.005):
    model     = RegNet(H)
    optimizer = torch.optim.Adam(model.parameters(), lr=lr)
    criterion = nn.MSELoss()

    for epoch in range(n_epochs):
        model.train()
        optimizer.zero_grad()
        loss = criterion(model(X_tensor), y_tensor)
        loss.backward()
        optimizer.step()

    return model

# Part (a): Train H = 1, 5, 20, 100 
hidden_sizes = [1, 5, 20, 100]
models = {}
for H in hidden_sizes:
    print(f"Training H={H}...")
    models[H] = train_regnet(H)

# Plot
fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, H in zip(axes.flatten(), hidden_sizes):
    models[H].eval()
    with torch.no_grad():
        y_pred = models[H](X_tensor).numpy()

    ax.scatter(X.ravel(), y, s=10, alpha=0.4, label='Noisy data')
    ax.plot(X.ravel(), y_true, 'g--', linewidth=2, label='True sine')
    ax.plot(X.ravel(), y_pred, 'r',   linewidth=2, label=f'H={H} fit')
    ax.set_title(f'H = {H} hidden neurons')
    ax.set_xlabel('x'); ax.set_ylabel('y')
    ax.legend()

plt.suptitle('Part (a): RegNet fits for different H', fontsize=14)
plt.tight_layout()
plt.show()
```
